```python
import boto3
from pathlib import Path
```

```python
s3 = boto3.client("s3")

bucket_name = "dsan6000-sl2420"
bucket_name
```

```python
s3.create_bucket(Bucket=bucket_name)

print(f"Created bucket: {bucket_name}")
```

```python
local_dir = Path("data")
parquet_files = sorted(local_dir.glob("*.parquet"))

print(f"Found {len(parquet_files)} parquet files")
```

```python
for file_path in parquet_files:
    s3_key = f"wikipedia-hourly/{file_path.name}"
    
    print(f"Uploading {file_path} -> s3://{bucket_name}/{s3_key}")
    
    s3.upload_file(
        str(file_path),
        bucket_name,
        s3_key
    )
```

```python
response = s3.list_objects_v2(
    Bucket=bucket_name,
    Prefix="wikipedia-hourly/"
)

uploaded_files = [
    obj["Key"]
    for obj in response.get("Contents", [])
    if obj["Key"].endswith(".parquet")
]

print(f"Uploaded {len(uploaded_files)} parquet files")

for key in uploaded_files[:5]:
    print(key)
```
