```python
import boto3
from pathlib import Path
```

```python
s3 = boto3.client("s3")

source_bucket = "dsan6000-wikipedia"
local_dir = Path("data")

local_dir.mkdir(exist_ok=True)
```

```python
response = s3.list_objects_v2(
    Bucket=source_bucket,
    Prefix="hourly_parquet/"
)

for obj in response.get("Contents", [])[:10]:
    print(obj["Key"])
```

```python
objects = response.get("Contents", [])

parquet_objects = [
    obj["Key"]
    for obj in objects
    if obj["Key"].endswith(".parquet")
]

len(parquet_objects)
```

```python
for key in parquet_objects:
    filename = Path(key).name
    local_path = local_dir / filename
    
    print(f"Downloading {key} -> {local_path}")
    s3.download_file(
        source_bucket,
        key,
        str(local_path)
    )
```

```python
downloaded_files = sorted(local_dir.glob("*.parquet"))

print(f"Downloaded {len(downloaded_files)} parquet files")

for path in downloaded_files[:5]:
    print(path)
```
